```python
import numpy as np
```

```python
subjects = ["Python", "Maths", "English", "Java", "WebDev"]

marks = np.random.randint(35, 100, (10, 5))

print("Student Marks:")
print(marks)
```

```python
for i in range(10):
    print("Student", i+1, ":", marks[i])
```

```python
average = np.mean(marks, axis=1)

print("Average Marks:")

for i in range(10):
    print("Student", i+1, ":", round(average[i], 2))
```

```python
for i in range(5):
    print(subjects[i], "Average:", round(np.mean(marks[:, i]), 2))
```

```python
topper = np.argmax(average)

print("Class Topper: Student", topper+1)
print("Highest Average:", average[topper])
```

```python
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
```

```python
passed = 0
failed = 0

for i in range(10):
    if average[i] >= 40:
        passed = passed + 1
    else:
        failed = failed + 1

print("Students Passed:", passed)
print("Students Failed:", failed)
```

```python
percentage = (passed / 10) * 100

print("Pass Percentage:", percentage, "%")
```

```python
print("STUDENT PERFORMANCE REPORT")
print("--------------------------")

print("Total Students:", 10)
print("Total Subjects:", 5)
print("Class Average:", round(np.mean(marks), 2))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
print("Students Passed:", passed)
print("Students Failed:", failed)
print("Pass Percentage:", percentage, "%")

print("--------------------------")
```
